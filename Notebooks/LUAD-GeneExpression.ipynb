## Dataset Overview

- This project uses the public `GSE43458` dataset from the `NCBI Gene Expression Omnibus (GEO)`.
- The dataset contains 110 samples measured using the Affymetrix Human Gene 1.0 ST Array.

| Item | Details |
|---|---|
| Dataset | GSE43458 |
| Organism | Homo sapiens |
| Platform | GPL6244 |
| Total samples | 110 |
| Tumor samples | 80 |
| Tumors from never-smokers | 40 |
| Tumors from smokers | 40 |
| Normal lung samples | 30 |
| Expression data | RMA-normalized and log2-transformed |

- The normal samples are paired with 30 of the never-smoker tumor cases. 
- I will check the sample metadata to confirm the patient matching before running the paired analysis.
- The main analysis will compare matched tumor and normal samples from never-smokers. 
- A second analysis will compare tumors from smokers and never-smokers.

<h2 style="margin:0;color: #0F172A;">
Analysis Workflow
</h2>
</div>


The analysis follows these steps:

1. Download the expression dataset and sample metadata from GEO.
2. Check the sample IDs and assign the Tumor and Normal groups.
3. Review the expression distributions and sample patterns.
4. Identify differentially expressed features using limma.
5. Map array probe IDs to gene symbols before selecting genes for network analysis.
6. Explore protein interactions and identify candidate hub genes.
7. Run functional enrichment analysis.
8. Review selected genes using external expression, survival, protein, and drug–gene databases.

<h2 style="margin:0;color: #0F172A;">
Quality Control

</h2>
</div>

Before comparing gene expression, I checked the expression distributions and overall sample patterns.

The GEO expression values are already RMA-normalized and log2-transformed, so this step focuses on checking the data rather than applying normalization again.

The following plots are used to review the dataset:

- **Boxplot:** compares expression distributions across samples.
- **Density plot:** shows the shape of the expression distributions.
- **PCA:** summarizes major patterns of variation across samples.

### Expression Distribution

- The boxplot helps identify samples with unusual medians or spreads.

![Expression distribution across samples](../Images/QualityControl/expression_boxplot.png)

**Interpretation**

- Compare the medians and spreads across samples.
- Check whether any samples have noticeably different distributions.
- Review unusual samples before deciding whether they should be excluded.

### Density Plot

- The density plot provides another view of the expression distributions.

![Density plot of expression values](../Images/QualityControl/expression_density.png)

**Interpretation**

- Check how much the distributions overlap.
- Look for samples with noticeably different distribution shapes.
- Similar distributions do not necessarily mean that the biological groups are the same.

### Principal Component Analysis

- PCA was used to explore the main patterns of variation across samples. The plot uses the 1,000 most variable probes.

![PCA of samples](../Images/QualityControl/pca_samples.png)

**Interpretation**

- Check whether samples form visible groups or show unusual patterns.
- Review the distribution of Normal, Tumor_NeverSmoker, and Tumor_Smoker samples.
- PCA is exploratory and does not, by itself, establish statistical significance.

### Differential Expression Results

- Differential expression analysis was performed using the limma package to compare LUAD tumor samples with normal lung tissue.
- The analysis tested 33,252 features. Based on the current analysis script, 319 features were classified as upregulated and 706 as downregulated. 
- The complete results include log fold changes, raw p-values, adjusted p-values, and gene symbols where available.
- The statistical filtering criteria should be considered when interpreting these results.

### Interpretation of Differential Expression Results

**Volcano Plot**

- The volcano plot summarizes the magnitude and statistical significance of differential expression between LUAD tumor and normal lung samples. 
- Positive log2 fold changes indicate higher expression in tumor samples, whereas negative values indicate lower expression. 
- The plot distinguishes upregulated, downregulated, and non-significant features using the selected statistical thresholds.

**MA Plot**

- The MA plot displays log2 fold changes against average expression levels. 
- Most features are centered around zero, while a subset shows larger positive or negative changes. 
- This visualization helps assess the distribution of expression changes across different expression levels.

**Heatmap**

- The heatmap displays the expression patterns of the top 50 ranked features across samples. 
- Hierarchical clustering provides an exploratory view of sample similarity and potential differences between tumor and normal groups. 
- The observed patterns should be interpreted alongside the differential expression statistics.

**Summary**

- The analysis tested 33,252 features and classified 319 as upregulated and 706 as downregulated according to the current analysis script. 
- These counts refer to analyzed features and should not automatically be interpreted as unique genes unless duplicate gene symbols have been addressed.

<h2 style="margin:0;color: #0F172A;">
Protein-Protein Interaction (PPI) Network Analysis and Hub Gene Identification
</h2>



>**Input:**  
>Top differentially expressed genes (DEGs) identified from limma analysis (gene symbols)

>**Tools:**  
>1. **STRING Database (v12)** for protein–protein interaction network construction  
>2. **Cytoscape (v3.10.4)** for network visualization and topology analysis  
>3. **cytoHubba plugin** for hub gene identification

>**Parameters (STRING):**  
>- Organism: Homo sapiens  
>- Minimum interaction score: 0.7 (High confidence)  
>- Active interaction sources: Experiments, Databases, Co-expression 

>**Parameters (Cytoscape/cytoHubba):**  
>- Hub ranking algorithm: Maximum Clique Centrality (MCC)  
>- Network topology analysis performed on STRING-derived PPI network

>**Steps:**
>1. DEG gene symbols uploaded to STRING database
>2. Protein–protein interaction network generated using high-confidence interactions
>3. STRING network exported as TSV file and imported into Cytoscape
>4. Network visualization and topology analysis performed
>5. cytoHubba used to rank highly connected proteins and identify hub genes

>**Output:**  
>- PPI network visualization  
>- Ranked list of hub genes based on MCC score  
>- Identification of potential key regulatory genes involved in lung adenocarcinoma  
>- BUB1 identified as one of the top-ranked hub genes

### Functional Enrichment Summary

Functional enrichment analysis consistently demonstrated that the identified hub genes are primarily involved in cell cycle regulation, mitotic progression, chromosome segregation, DNA replication, and cell division.

Gene Ontology (GO) biological process analysis revealed significant enrichment of pathways associated with nuclear division, mitotic cell cycle, and chromosome organization. Similarly, KEGG pathway analysis highlighted the Cell Cycle pathway as the most significantly enriched pathway, while Reactome and WikiPathways further confirmed the involvement of mitotic checkpoints and DNA replication.

These findings indicate that dysregulation of cell-cycle-related processes is a fundamental molecular characteristic of lung adenocarcinoma. The consistent enrichment across multiple biological databases strengthens the reliability of the identified hub genes and supports their potential role as diagnostic biomarkers and therapeutic targets in LUAD.

>UALCAN web portal

>TCGA Lung Adenocarcinoma (LUAD)


>**Parameters:**

<h2 style="margin:0;color:#0F172A;">
Limitations
</h2>

</div>


- This analysis is based solely on a single public microarray dataset (GSE43458), which may not fully represent the molecular heterogeneity of LUAD across different populations and clinical contexts.

- The differentially expressed genes were identified at the transcriptomic level only, without integration with other molecular layers such as genomics, proteomics, or epigenomics (DNA methylation, chromatin accessibility).

- All downstream analyses—including functional enrichment, protein-protein interaction networks, and survival assessment—were conducted using publicly available databases (Enrichr, STRING, HPA, KM Plotter). These analyses are in silico and lack experimental or functional validation (e.g., qPCR, Western blot, CRISPR/RNAi).

- Clinical covariates such as age, sex, tumor stage, smoking history, treatment regimen, and overall survival status were not included in the statistical modeling, which may introduce confounding effects and limit the clinical generalizability of the findings.
